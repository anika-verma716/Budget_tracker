```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('expense_data_1.csv')

# Display success message
print("Dataset loaded successfully!")

# Display first 5 rows
display(df.head())

# Display column names
print("\nDataset Columns:")
print(df.columns.tolist())

# Display total rows and columns
print("\nDataset Shape:")
print(df.shape)
```

```python
# STEP 4: DATA CLEANING

# Create a copy of the original dataset
df_clean = df.copy()

# Remove unnecessary columns
df_clean = df_clean.drop(
    columns=['Subcategory', 'Note.1', 'Account.1', 'INR'],
    errors='ignore'
)

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Convert Date into datetime format
df_clean['Date'] = pd.to_datetime(
    df_clean['Date'],
    format='%m/%d/%Y %H:%M',
    errors='coerce'
)

# Convert Amount into numeric format
df_clean['Amount'] = pd.to_numeric(
    df_clean['Amount'],
    errors='coerce'
)

# Remove records with missing essential values
df_clean = df_clean.dropna(
    subset=['Date', 'Amount', 'Category', 'Income/Expense']
)

# Fill missing descriptions
df_clean['Note'] = df_clean['Note'].fillna('No Description')

# Remove unnecessary spaces
df_clean['Category'] = df_clean['Category'].str.strip()

df_clean['Income/Expense'] = (
    df_clean['Income/Expense'].str.strip()
)

# Display cleaned dataset
print("DATA CLEANING COMPLETED SUCCESSFULLY!")

print("\nCleaned Dataset:")
display(df_clean.head())

print("\nDataset Shape After Cleaning:")
print(df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum())
```

```python
# STEP 5: FINANCIAL SUMMARY

# Calculate total income
total_income = df_clean[
    df_clean['Income/Expense'] == 'Income'
]['Amount'].sum()

# Calculate total expenses
total_expenses = df_clean[
    df_clean['Income/Expense'] == 'Expense'
]['Amount'].sum()

# Calculate net savings
savings = total_income - total_expenses

# Display financial summary
print("\n========== FINANCIAL SUMMARY ==========")

print("Total Income: Rs.", round(total_income, 2))

print("Total Expenses: Rs.", round(total_expenses, 2))

print("Net Savings: Rs.", round(savings, 2))

if savings >= 0:
    print("Financial Status: Positive Balance")
else:
    print("Financial Status: Expenses Exceed Income")

print("=======================================")
```

```python
# STEP 6: CATEGORY-WISE EXPENSE ANALYSIS

# Select expense transactions
expenses = df_clean[
    df_clean['Income/Expense'] == 'Expense'
]

# Calculate category-wise expenses
category_expenses = expenses.groupby(
    'Category'
)['Amount'].sum().sort_values(ascending=False)

# Display results
print("\n========== CATEGORY-WISE EXPENSES ==========")

print(category_expenses.to_string())

# Find highest spending category
highest_category = category_expenses.idxmax()

highest_amount = category_expenses.max()

print("\nHighest Spending Category:", highest_category)

print("Amount Spent: Rs.", round(highest_amount, 2))

print("============================================")
```

```python
# STEP 7: MONTHLY EXPENSE ANALYSIS

# Select expense records
expenses = df_clean[
    df_clean['Income/Expense'] == 'Expense'
].copy()

# Extract month and year
expenses['Month'] = expenses['Date'].dt.to_period('M')

# Calculate monthly expenses
monthly_expenses = expenses.groupby(
    'Month'
)['Amount'].sum().sort_index()

# Display results
print("\n========== MONTHLY EXPENSE ANALYSIS ==========")

print(monthly_expenses.to_string())

# Find highest spending month
if not monthly_expenses.empty:

    highest_month = monthly_expenses.idxmax()

    print("\nHighest Spending Month:", highest_month)

    print(
        "Total Expenses: Rs.",
        round(monthly_expenses.max(), 2)
    )

print("==============================================")
```

```python
# STEP 8: MONTHLY BUDGET MANAGEMENT

def budget_management():

    print("\n========== BUDGET MANAGEMENT ==========")

    try:

        # Enter monthly budget
        budget = float(
            input("Enter your monthly budget (Rs.): ")
        )

        if budget < 0:
            print("Budget cannot be negative!")
            return

        # Select month
        month = input(
            "Enter month (YYYY-MM): "
        )

        selected_month = pd.Period(
            month, freq='M'
        )

        # Calculate expenses for selected month
        monthly_expenses = df_clean[
            (df_clean['Income/Expense'] == 'Expense') &
            (df_clean['Date'].dt.to_period('M') == selected_month)
        ]['Amount'].sum()

        # Calculate remaining budget
        remaining_budget = budget - monthly_expenses

        # Display results
        print("\n========== BUDGET REPORT ==========")

        print("Selected Month:", selected_month)

        print("Monthly Budget: Rs.", budget)

        print(
            "Total Expenses: Rs.",
            round(monthly_expenses, 2)
        )

        print(
            "Remaining Budget: Rs.",
            round(remaining_budget, 2)
        )

        # Budget status
        if remaining_budget < 0:

            print("\nWARNING! You have exceeded your budget.")

            print(
                "Extra Spending: Rs.",
                round(abs(remaining_budget), 2)
            )

        else:

            print("\nGreat! You are within your budget.")

        print("===================================")

    except ValueError:

        print("Invalid input! Please try again.")


# Run budget management
budget_management()
```

```python
df_clean.to_csv('budget_tracker_report.csv', index=False)

print("Budget report exported successfully!")
print("File name: budget_tracker_report.csv")
```
